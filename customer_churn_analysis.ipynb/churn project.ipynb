```python
#5 How should missing values in the age column be handled
df["age"].fillna(df["age"].mean(),inplace = True)
df.isna().sum()
```

```python
# 6 How do you handle missing values in the city column
df["city"].mode()[0]
df["city"].value_counts()
df["city"].fillna(df["city"].mode()[0],inplace = True)
```

```python
# 7 How to treat missing values in avg_session_time?
df["avg_session_time"].fillna(df["avg_session_time"].mean(),inplace = True)
df["avg_session_time"].value_counts()
```

```python
# 8 How to handle missing values in discount_percent
df["discount_percent"].fillna(0,inplace = True)
```

```python
# 10  How do you remove duplicate rows?
df.drop_duplicates()
```

```python
df.dtypes
```

```python
# 11 Are date columns in proper datetime format?
df["signup_date"]= pd.to_datetime(df["signup_date"])
df["last_active_date"] = pd.to_datetime(df["last_active_date"])
df.dtypes
```

```python
# 12 Are there any illogical date values (last active before signup)?

df[df["last_active_date"] < df["signup_date"]]
```

```python
# 13 How do you handle illogical date records? Check if there are any records where 
# (Last active date is < than sign up date)

df1 = df[df["last_active_date"] >= df["signup_date"]]
df1
```

```python
df1
```

```python
# 14 Are there invalid age values? (Age> 80, Age<18)
df1[(df1["age"] < 18) | (df1["age"] > 80)]
```

```python
# Create a customer_tenure_years column (Calculate Years from Tenure Months)
df1["customer_tenure_year"] = df1["tenure_months"] / 12
df1
```

```python
# 16. Create a revenue_per_month column (Create a column Total rev / Tenure months)
df1["revenue_per_month"] = df1["total_revenue"]/df1["tenure_months"]
df1
```

```python
# 17. Create a binary churn flag (1 is Churn is “Yes”, else 0)
df1["binary_churn"] = df1["churned"].apply(lambda x : 1 if x == "Yes" else 0)
df1
```

```python
# 18. Create a customer age group column (Age < 25 -> “Young”, <45 -> “Middle”, else
 # “Senior”)
df1["age_group"] = df1["age"].apply(lambda x : "young" if x <= 25 else("middle" if x <= 45 else "senior"))
df1
```

```python
# 19. Create a high-value customer flag 
# (If total revenue > Median of total revenue -> Yes or else No)
def abcd_(total_revenue):
    return "Yes" if total_revenue > df1["total_revenue"].median() else "NO"

df1["high-vc"] = df1["total_revenue"].apply(abcd_)
df1
```

```python
# 20. Create a days_active column (Last active date – Sign up date)

df1["days_active"] = df1 ["last_active_date"]-df1["signup_date"]
df1
```

```python
# 21. Create session usage category(Avg_session_time < 30 → Low,
# Avg_session_time < 60 → Medium , else “High”)

df1["suc"]=df1["avg_session_time"].apply(lambda x : "low" if x < 30 else("medium" if x < 60 else "high"))
df1
```

```python
# 22. Create discount category (Discount = 0 → “No Discount”,
# Discount <= 10 → “Low Discount”, else “High Discount”)
df1["dc"]= df1["discount_percent"].apply(lambda x : "no_discount" if x == 0 else ("low" if x <= 10 else "high"))
df1
```

```python
#23. Extract signup month
# 24. Extract signup year

df1["month_"] = df1["signup_date"].dt.month
df1["year_"] = df1["signup_date"].dt.year

df1
```

```python
# 25. Create a support_intensity feature (If support_tickets >= 5 → “High”, else “Low”)
df1["support_intensity"]= df1["support_tickets"].apply(lambda x : "high" if x > 5 else "low")
df1
```

```python
#26 . Create revenue category (Revenue < 10000 → “Low”, 
# Revenue < 30000 → “Medium”,else → “High”)
df1["revenue_cat"]= df1["total_revenue"].apply(lambda x : "low" if x < 10000 else ("medium" if x <= 30000 else "high"))
df1
```

```python
# 27. Create plan rank for modeling (Plan type “Basic” → 1, “Standard” → 2, “Premium” → 3)

df1["plan_rank"]= df1["plan_type"].apply(lambda x : 1 if x == "Basic" else(2 if x ==  "Standard" else 3))
df1
```

```python
# 28. Create engagement score
df1["engagement_score"] = df1["avg_session_time"]/(df1["support_tickets"]+1)
df1
```
